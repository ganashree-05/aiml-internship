```python
def mean(array):
    n=len(array)
    return np.sum(array)/n
```

```python
mean(np.array([1,2,3,4,5,6]))
```

```python
def variance(array):
    n=len(array)
    m=mean(array)
    return np.sum((array - m)**2)/n
```

```python
variance(np.array([1,2,3,4,5,6]))
```

```python
import numpy as np
scores = np.random.randint(50, 101, size=(5, 3))
mean_scores = scores.mean(axis=0)
centered_scores = scores - mean_scores
print("Original Scores (5 students x 3 subjects):")
print(scores)

print("\nMean Score of Each Subject:")
print(mean_scores)

print("\nCentered Scores (After Normalization):")
print(centered_scores)
```
