```python
Define the Sample Space 
A customer can perform one of three actions:
Click (C)
Scroll (S)
Exit (E)
For two consecutive actions, we consider all ordered pairs.
So the sample space S is:
S={(C,C),(C,S),(C,E),(S,C),(S,S),(S,E),(E,C),(E,S),(E,E)}
Total possible outcomes:
     ∣S∣=3×3=9
Probability of Event E: "Customer clicks at least once"
Event definition:
Customer performs Click at least once in the two actions.
Outcomes where Click appears:
(C, C)
(C, S)
(C, E)
(S, C)
(E, C)
So, 
    ∣E∣=5
Total outcomes = 9
    P(E)=5/9
Final Answer:
  P(E)=5/9=0.556
```

```python
Python Simulation: Rolling Two Dice 1000 Times
We now simulate rolling two dice 1000 times and compute the probability that the sum = 7.
Theoretical Probability
Possible outcomes for sum = 7:
(1,6), (2,5), (3,4), (4,3), (5,2), (6,1)
Total favorable = 6
Total possible outcomes = 36
   P(sum = 7)=6/36=1/6=0.1667
```

```python
import numpy as np
n = 100000
count = 0
for _ in range(n):
    bag = np.array(["R"] * 5 + ["B"] * 5)
    np.random.shuffle(bag)
    first, second = bag[0], bag[1]
    if first == "R" and second == "R":
        count += 1

probability = count / n

print("Experimental Probability:", probability)
```

```python
Given:

P(Spam) = 0.1  
P(Ham) = 0.9  

P(Free | Spam) = 0.9  
P(Free | Ham) = 0.05  

Using Total Probability:

P(Free) = P(Free|Spam)P(Spam) + P(Free|Ham)P(Ham)

= (0.9 × 0.1) + (0.05 × 0.9)
= 0.09 + 0.045
= 0.135

Using Bayes' Theorem:

P(Spam | Free) = (P(Free|Spam) × P(Spam)) / P(Free)

= (0.9 × 0.1) / 0.135
= 0.667

Conclusion:
If an email contains the word "Free", there is a 66.7% chance it is Spam.
```
