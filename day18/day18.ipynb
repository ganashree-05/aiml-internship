```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("internship.db")
cursor = conn.cursor()

cursor.execute("CREATE TABLE IF NOT EXISTS mentors (mentor_id INTEGER, mentor_name TEXT, track TEXT)")

cursor.execute("INSERT INTO mentors VALUES (1, 'Dr. Sharma', 'Data Science')")
cursor.execute("INSERT INTO mentors VALUES (2, 'Mr. Rao', 'Web Dev')")
cursor.execute("INSERT INTO mentors VALUES (3, 'Ms. Priya', 'AI/ML')")
cursor.execute("INSERT INTO mentors VALUES (4, 'Mr. Khan', 'Cyber Security')")

conn.commit()

query = """
SELECT interns.name, interns.track, mentors.mentor_name
FROM interns
INNER JOIN mentors
ON interns.track = mentors.track
"""

df = pd.read_sql_query(query, conn)

print(df)

conn.close()
```

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("internship.db")
```

```python
query1 = """
SELECT * FROM interns
WHERE track = 'Data Science' AND stipend > 5000
"""

df1 = pd.read_sql_query(query1, conn)
print(df1)
```

```python
query2 = """
SELECT track, AVG(stipend) AS avg_stipend
FROM interns
GROUP BY track
"""

df2 = pd.read_sql_query(query2, conn)
print(df2)
```

```python
query3 = """
SELECT track, COUNT(*) AS total_interns
FROM interns
GROUP BY track
"""

df3 = pd.read_sql_query(query3, conn)
print(df3)

conn.close()
```
